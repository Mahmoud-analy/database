```python
# أعمدة الإصابات/الوفيات: الفراغ معناه = صفر (لم يحدث)
injury_cols = ['Operator Employee Injuries', 'Operator Contractor Injuries',
               'Emergency Responder Injuries', 'Other Injuries', 'Public Injuries', 'All Injuries']
fatality_cols = ['Operator Employee Fatalities', 'Operator Contractor Fatalities',
                  'Emergency Responder Fatalities', 'Other Fatalities', 'Public Fatalities', 'All Fatalities']

df[injury_cols] = df[injury_cols].fillna(0)
df[fatality_cols] = df[fatality_cols].fillna(0)

# أعمدة التكلفة: القيم الفاضية قليلة جدًا (6-16 من 2795)، ممكن نملأها بصفر أو نحذف الصفوف
cost_cols = ['Property Damage Costs', 'Lost Commodity Costs', 'Public/Private Property Damage Costs',
             'Emergency Response Costs', 'Environmental Remediation Costs', 'Other Costs']
df[cost_cols] = df[cost_cols].fillna(0)
print(df[injury_cols + fatality_cols + cost_cols].isnull().sum())

```
